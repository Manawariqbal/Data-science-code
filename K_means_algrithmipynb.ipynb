```python
#import the libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
#




X, y_true = make_blobs(n_samples=500, centers=3, cluster_std=4, random_state=42)
```

```python
df=pd.DataFrame(X,columns=['Feature_1','Feature_2'])
df.head()
```

```python
#Scaling
scaler=StandardScaler()
X_scaled=scaler.fit_transform(df)

```

```python
inertia=[]
k_range=range(1,11)
for k in k_range:
  kmeans=KMeans(n_clusters=k, random_state=42)
  kmeans.fit(X_scaled)
  inertia.append(kmeans.inertia_)
inertia
```

```python
plt.plot(k_range,inertia,marker='o')
```

```python
kmeans_final=KMeans(n_clusters=3,random_state=42)
#Train

cluster_labels=kmeans_final.fit_predict(X_scaled)


df['cluster']=cluster_labels
sns.scatterplot(
    x=df['Feature_1'],
    y=df['Feature_2'],
    hue=df['cluster'],

    palette='viridis'
)
```

```python
#DBScan Algorithm
from sklearn.datasets import  make_moons
X,y_true=make_moons(n_samples=500,noise=0.05,random_state=42)
```

```python
from sklearn.cluster import KMeans,DBSCAN
```

```python
df=pd.DataFrame(X,
                columns=['Feature_1','Feature_2'])
df.head()
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(df)
kmeans=KMeans(n_clusters=2,random_state=42)

#Train
Kmeans_lable=kmeans.fit_predict(X_scaled)


```

```python
df['kmean_cluster']=Kmeans_lable
sns.scatterplot(
    x=df['Feature_1'],
    y=df['Feature_2'],
    hue=df['kmean_cluster'],
    palette='tab10'
)
```

```python
#Using DB scan
dbscan=DBSCAN(eps=0.3,min_samples=5)
dbscan_label=dbscan.fit_predict(X_scaled)


df['dbscan_cluster']=dbscan_label

sns.scatterplot(
    x=df['Feature_1'],
    y=df['Feature_2'],
    hue=df['dbscan_cluster'],
    palette='tab10'
)
```
