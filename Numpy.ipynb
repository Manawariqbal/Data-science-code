```python
#Given an array of student marks, calculate mean, median, standard deviation, minimum, and maximum.

marks=np.array([80,98,72,34,67,89,76,67,90,78,89])

marks

#mean
print("Mean=: ",marks.mean())

#median
print("Median: ",np.median(marks))


#standard deviation

print("Standard deviation: ",np.std(marks))


#minimum
print("Minimum",np.min(marks))


#maximum
print("Maximum ",np.max(marks))
```
