```python
#constructor -It's method which is called automatically when an object is created for the class

#__init__()
class Student:
  #constructor
  def __init__(self,name,age):
    self.name=name
    self.age=age


#self refer to the current object

s1=Student("Rahul",24)
s2=Student("Aman",23)
```

```python
#instance variable
#attribute belong to the object
class Employee:
  #class variable
  #Attribute belonging to the class

  company="ABC LTD"
  def __init__(self,name,salary):
    self.name=name
    self.salary=salary


  #instance method
  #method belongs to the object

  def display(self):
    print("Name: ",self.name)
    print("Salary: ",self.salary)

e1=Employee("Amir",300000)
e1.display()
print(Employee.company)
```

```python
#Student
#name,age,class
#School name

#method
#display info
class Student:

    school_name="ABC School"

    def __init__(self, name, age, student_class):

        self.name=name

        self.age=age

        self.student_class=student_class



    def display_info(self):

        print("Name:", self.name)

        print("Age:", self.age)

        print("Class:", self.student_class)

        print("School Name:", Student.school_name)






student1 = Student("Pooja", 20, "Diploma")



student1.display_info()
```

```python
#inheritance
#inheritance allow a class to acquired properties and method from another class


class Animal:

  def eat(self):
    print("The animal is eating")


class Dog(Animal):

  def bark(self):
    print("Dog is barking")


dog1=Dog()
dog1.eat()
dog1.bark()
```

```python
class Person:
  def __init__(self,name):
    self.name=name
  def introduce(self):
    print("Hi my name is ",self.name)

class Student(Person):


  def study(self):
    print(self.name ,"is studying")



class Employee(Person):

  def work(self):
    print(self.name,"is working ")



s1=Student("Aman")
e1=Employee("Neha")
e1.introduce()
e1.work()
s1.introduce()
s1.study()


```

```python
#Method overridng

#Child class override the parent class method
class Animal:
  def sound(self):
    print("Animal is making sound ")
class Dog(Animal):
  def sound(self):
    print("The dog is barking ")
class Cat(Animal):
  def sound(self):
    print("The cat is meowing ")

e=Animal()
d=Dog()
c=Cat()
print(e.sound())
print(d.sound())
print(c.sound())

```
